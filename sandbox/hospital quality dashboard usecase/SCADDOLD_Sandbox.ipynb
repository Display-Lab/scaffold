## Create a .env file that points to the sandbox knowledge base
SCAFFOLD requires an environment file to be set up before running the pipeline. For more details on how to configure this file and available setup options, see the main [SCAFFOLD README](https://github.com/Display-Lab/scaffold/blob/main/README.md).
The following code cell creates a `.env` configuration specifically for running SCAFFOLD with this sandbox use case.

```python
%%writefile .env
default_preferences=file:///content/scaffold/sandbox/hospital quality dashboard usecase/knowledge-base/preferences.json
mpm=/content/scaffold/sandbox/hospital quality dashboard usecase/knowledge-base/prioritization_algorithms/motivational_potential_model.csv
manifest=file:///content/scaffold/sandbox/hospital quality dashboard usecase/knowledge-base/manifest.yaml
config=file:///content/scaffold/sandbox/hospital quality dashboard usecase/knowledge-base/config.yaml
preferences=/content/scaffold/sandbox/hospital quality dashboard usecase/data/tabular inputs/Preference.csv
history=/content/scaffold/sandbox/hospital quality dashboard usecase/data/tabular inputs/MessageHistory.csv
log_level=DEBUG
meas_period=1
min_count=10
generate_image=1
```

## Run scaffold with the .env file and path to the sandbox data
The following command runs the `batch-csv` command-line interface from SCAFFOLD, sets the environment file path (`ENV_PATH`), provides the path to the tabular input data for the Hospital Quality Dashboard use case, and sets the performance month.
For details on running other commands and configuring additional command parameters, refer to the main [SCAFFOLD README](https://github.com/Display-Lab/scaffold/blob/main/README.md).

```python
! ENV_PATH=./.env pipeline  batch-csv scaffold/sandbox/hospital\ quality\ dashboard\ usecase/data/tabular\ inputs --performance-month 2025-01-01
```
